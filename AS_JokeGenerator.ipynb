Before you can use the OpenAI API, you'll need an API key. If you don't already have one, create one on the [OpenAI website](https://platform.openai.com/account/api-keys).

In Colab, add the key to the secrets manager under the "🔑" icon in the left panel. Give it the name `OPENAI_API_KEY`. Then, we'll retrieve it and configure the OpenAI client.

```python
from openai import OpenAI
from google.colab import userdata

# Retrieve the API key from Colab secrets
OPENAI_API_KEY = userdata.get('OPENAI_API_KEY')

# Initialize the OpenAI client
client = OpenAI(api_key=OPENAI_API_KEY)

print("OpenAI client initialized successfully!")
```

Now, let's define a function that will use the OpenAI model to generate a joke based on your input word. We'll use the `gpt-4o` model for this task.

```python
def generate_joke(word):
    try:
        response = client.chat.completions.create(
            model="gpt-4o", # Using a new, powerful OpenAI model
            messages=[
                {"role": "system", "content": "You are a witty comedian. Generate a short, clean, and funny joke related to the user's input word."},
                {"role": "user", "content": f"Tell me a joke about '{word}'."}
            ],
            max_tokens=150
        )
        return response.choices[0].message.content.strip()
    except Exception as e:
        return f"Error generating joke: {e}"

print("Joke generation function defined.")
```

Finally, we'll create an interactive user interface using `ipywidgets` so you can easily enter a word and get a joke.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create an input text box
word_input = widgets.Text(
    value='',
    placeholder="Enter a word here (e.g., 'chicken')",
    description='Your Word:',
    disabled=False
)

# Create a button
button = widgets.Button(description="Generate Joke")

# Create an output widget to display the joke
output_widget = widgets.Output()

def on_button_click(b):
    with output_widget:
        clear_output()
        word = word_input.value
        if word:
            print(f"Generating joke for '{word}'...")
            joke = generate_joke(word)
            print("\n--- Your Joke ---\n")
            print(joke)
        else:
            print("Please enter a word to generate a joke.")

# Attach the button click event to the function
button.on_click(on_button_click)

# Display the widgets
display(word_input, button, output_widget)

print("Joke generator tool ready! Enter a word and click 'Generate Joke'.")
```

Now, let's create a Gradio interface for the `generate_joke` function. We'll define the input and output components and then launch the app.

```python
import gradio as gr

# Create the Gradio interface
iface = gr.Interface(
    fn=generate_joke, # The function to wrap
    inputs=gr.Textbox(lines=1, placeholder="Enter a word..."), # Input component
    outputs="text", # Output component
    title="AI Joke Generator",
    description="Enter a word and let the AI generate a funny joke for you!",
    allow_flagging="never"
)

# Launch the Gradio app
print("Launching Gradio app...")
iface.launch(share=True, debug=True)
```
